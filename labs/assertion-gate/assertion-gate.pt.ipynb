# Por que esta resposta bloqueou a liberação?

Execute as células a partir da pasta do lab. Compare duas respostas sobre reembolso com uma política e corrija a que falhou.

A política informa **5 dias úteis após a aprovação**. A resposta informa **14 dias úteis**. No exemplo fornecido, o programa não encontra apoio para 14 e bloqueia a liberação (`HOLD`). Faça a alteração abaixo e compare os resultados.

```python
import json
import subprocess
import sys
from pathlib import Path
from tempfile import TemporaryDirectory

lab = Path.cwd() / "assertion_gate.py"
samples = lab.parent / "samples" / "claims.jsonl"
command = [sys.executable, str(lab), "--lang", "pt"]
before = subprocess.run(command, check=True, capture_output=True, text=True).stdout
print(before, end="")
assert "-> HOLD" in before
```

### Compare com a resolução

Use o prazo de **5**. O avaliador deve aprovar o caso (`PROMOTE`). A condição **após a aprovação** continua ausente da resposta.

```python
rows = [json.loads(line) for line in samples.read_text(encoding="utf-8").splitlines()]
failing_case = next(row for row in rows if row["id"] == "refund-14d")
failing_case["claim"] = failing_case["claim"].replace("14", "5")

with TemporaryDirectory(prefix="lumina-lab-") as folder:
    candidate = Path(folder) / "claims.jsonl"
    candidate.write_text("\n".join(json.dumps(row) for row in rows) + "\n", encoding="utf-8")
    after = subprocess.run(
        [*command, "--input", str(candidate), "--ci"], check=True, capture_output=True, text=True
    ).stdout

print(after, end="")
assert "-> PROMOTE" in after
```
